# AI Armenian Font Creation

Prototype workflow to create an Armenian TTF from (1) an existing font, (2) a text style prompt, or (3) a reference image. Prompt/reference modes use a diffusion model to make glyph images; the notebook traces those images into a simple vector font. Image models often render Armenian characters inaccurately, so inspect the glyph contact sheet and treat the result as an experimental starting point, not production typography.

Run in Google Colab or a local GPU runtime for prompt/reference generation. Use a font whose license permits modification and redistribution. Model downloads and generation can require substantial disk space and VRAM.

```python
from pathlib import Path
import io, math, os, requests, torch, cv2
import numpy as np
from PIL import Image, ImageDraw, ImageFont, ImageOps
import matplotlib.pyplot as plt

# Armenian alphabet, upper and lowercase, plus common punctuation and space.
ARMENIAN = list('ԱԲԳԴԵԶԷԸԹԺԻԼԽԾԿՀՁՂՃՄՅՆՇՈՉՊՋՌՍՎՏՐՑՒՓՔՕՖ'
                 'աբգդեզէըթժիլխծկհձղճմյնշոչպջռսվտրցւփքօֆ')
CHARACTERS = list(dict.fromkeys([' '] + ARMENIAN + ['։', '՝', '՛', '՞', '՜', '՚', '֊', 'և']))
OUT_DIR = Path('armenian_font_output')
OUT_DIR.mkdir(exist_ok=True)
print(f'{len(CHARACTERS)} glyphs to create; output: {OUT_DIR.resolve()}')
```

## Choose an input

`font` mode converts the supplied font's Armenian glyphs directly (good for testing the font-building stage). `prompt` creates glyphs from a text description. `reference` uses an image as visual conditioning. For model modes, each character is generated separately to reduce character mixing, but the model may still draw the wrong letter.

```python
MODE = 'font'  # 'font', 'prompt', or 'reference'
STYLE_PROMPT = 'elegant Armenian calligraphic lettering, consistent ink strokes, black glyph on plain white background'
REFERENCE_IMAGE = None  # e.g. 'reference.png'; used only when MODE='reference'
SOURCE_FONT = None  # e.g. '/path/to/Armenian-font.ttf'; used only when MODE='font'
FONT_FAMILY = 'Armenian AI Prototype'
FONT_FILENAME = OUT_DIR / 'armenian_ai_prototype.ttf'
IMAGE_SIZE = 512

# A freely licensed Noto Sans Armenian fallback for the demo. Check the license
# of any font you substitute before redistributing the resulting font.
FALLBACK_FONT_URL = 'https://github.com/google/fonts/raw/main/ofl/notosansarmenian/NotoSansArmenian%5Bwght%5D.ttf'
fallback_path = OUT_DIR / 'NotoSansArmenian.ttf'
if MODE == 'font' and SOURCE_FONT is None:
    if not fallback_path.exists():
        response = requests.get(FALLBACK_FONT_URL, timeout=60)
        response.raise_for_status()
        fallback_path.write_bytes(response.content)
    SOURCE_FONT = str(fallback_path)

assert MODE in {'font', 'prompt', 'reference'}
if MODE == 'reference':
    assert REFERENCE_IMAGE and Path(REFERENCE_IMAGE).is_file(), 'Set REFERENCE_IMAGE to an image file.'
if MODE == 'font':
    assert Path(SOURCE_FONT).is_file(), f'Font not found: {SOURCE_FONT}'
```

## Generate or render glyph images

The diffusion pipeline is initialized only for prompt/reference modes. You can set `MODEL_ID` to another compatible Diffusers text-to-image or image-to-image model. Review the model card and comply with its license and usage restrictions.

```python
MODEL_ID = 'stabilityai/stable-diffusion-xl-base-1.0'
device = 'cuda' if torch.cuda.is_available() else 'cpu'
glyph_images = {}

if MODE == 'font':
    source = ImageFont.truetype(SOURCE_FONT, 360)
    for ch in CHARACTERS:
        canvas = Image.new('L', (IMAGE_SIZE, IMAGE_SIZE), 255)
        if ch != ' ':
            draw = ImageDraw.Draw(canvas)
            box = draw.textbbox((0, 0), ch, font=source)
            w, h = box[2] - box[0], box[3] - box[1]
            draw.text(((IMAGE_SIZE-w)/2-box[0], (IMAGE_SIZE-h)/2-box[1]), ch, font=source, fill=0)
        glyph_images[ch] = canvas
else:
    from diffusers import StableDiffusionXLPipeline, StableDiffusionXLImg2ImgPipeline
    dtype = torch.float16 if device == 'cuda' else torch.float32
    if MODE == 'prompt':
        pipe = StableDiffusionXLPipeline.from_pretrained(MODEL_ID, torch_dtype=dtype, use_safetensors=True)
    else:
        pipe = StableDiffusionXLImg2ImgPipeline.from_pretrained(MODEL_ID, torch_dtype=dtype, use_safetensors=True)
    pipe = pipe.to(device)
    ref = Image.open(REFERENCE_IMAGE).convert('RGB').resize((IMAGE_SIZE, IMAGE_SIZE)) if MODE == 'reference' else None
    for index, ch in enumerate(CHARACTERS):
        if ch == ' ':
            glyph_images[ch] = Image.new('L', (IMAGE_SIZE, IMAGE_SIZE), 255)
            continue
        # The model is asked to draw exactly one character. Verify its output manually.
        prompt = (f'{STYLE_PROMPT}. One single Armenian letter {ch}, centered, isolated, '
                  'complete black glyph, clean white background, no other symbols or text.')
        seed = torch.Generator(device=device).manual_seed(4100 + index)
        if MODE == 'prompt':
            result = pipe(prompt=prompt, negative_prompt='multiple letters, words, watermark, frame',
                          width=IMAGE_SIZE, height=IMAGE_SIZE, num_inference_steps=30,
                          guidance_scale=7.0, generator=seed).images[0]
        else:
            result = pipe(prompt=prompt, image=ref, strength=0.55, num_inference_steps=30,
                          guidance_scale=7.0, generator=seed).images[0]
        glyph_images[ch] = result.convert('L')
    del pipe
    if device == 'cuda': torch.cuda.empty_cache()

print('Rendered', len(glyph_images), 'glyph images')
```

## Inspect and convert images to font outlines

The conversion thresholds each image, finds contours, and builds a TrueType font using straight-line outlines. This simple tracing can lose fine details; edit the glyph images or use a dedicated vectorization/type-design workflow for better results.

```python
def show_contact_sheet(images, chars, columns=8):
    rows = math.ceil(len(chars) / columns)
    fig, axes = plt.subplots(rows, columns, figsize=(columns*1.5, rows*1.65))
    axes = np.array(axes).reshape(-1)
    for ax, ch in zip(axes, chars):
        ax.imshow(images[ch], cmap='gray', vmin=0, vmax=255)
        ax.set_title('space' if ch == ' ' else ch)
        ax.axis('off')
    for ax in axes[len(chars):]: ax.axis('off')
    plt.tight_layout(); plt.show()

show_contact_sheet(glyph_images, CHARACTERS)
```

```python
def to_binary_mask(image):
    arr = np.asarray(image.convert('L'))
    # Infer foreground polarity from the image border, then Otsu-threshold.
    border = np.concatenate([arr[0], arr[-1], arr[:, 0], arr[:, -1]])
    if np.median(border) < 127: arr = 255 - arr
    _, mask = cv2.threshold(arr, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
    return mask

def glyph_outline(image, pen, units=1000, max_height=700, max_width=650):
    mask = to_binary_mask(image)
    if cv2.countNonZero(mask) < 8: return
    contours, hierarchy = cv2.findContours(mask, cv2.RETR_CCOMP, cv2.CHAIN_APPROX_SIMPLE)
    if hierarchy is None: return
    hierarchy = hierarchy[0]
    x, y, w, h = cv2.boundingRect(mask)
    scale = min(max_height / max(h, 1), max_width / max(w, 1))
    xoff = (units - w*scale) / 2
    yoff = (max_height - h*scale) / 2
    for i, contour in enumerate(contours):
        pts = cv2.approxPolyDP(contour, epsilon=1.2, closed=True).reshape(-1, 2)
        if len(pts) < 3: continue
        # In font coordinates outer rings are clockwise and holes counter-clockwise.
        area = sum(pts[j,0]*pts[(j+1)%len(pts),1] - pts[(j+1)%len(pts),0]*pts[j,1]
                   for j in range(len(pts))) / 2
        is_hole = hierarchy[i][3] != -1
        want_negative = not is_hole
        if (area < 0) != want_negative: pts = pts[::-1]
        coords = [(round(xoff+(px-x)*scale), round(yoff+(h-(py-y))*scale)) for px,py in pts]
        pen.moveTo(coords[0])
        for point in coords[1:]: pen.lineTo(point)
        pen.closePath()

# Save generated images so they can be inspected/edited and reused.
for i, (ch, image) in enumerate(glyph_images.items()):
    image.save(OUT_DIR / f'glyph_{ord(ch):04X}.png')
print('Saved glyph PNGs to', OUT_DIR)
```

## Build and save the TTF

The generated font uses Unicode code points for the selected Armenian characters, plus a basic `.notdef` glyph. Validate the font in multiple applications and test Armenian shaping; this prototype does not implement advanced hinting or kerning.

```python
from fontTools.fontBuilder import FontBuilder
from fontTools.pens.ttGlyphPen import TTGlyphPen

units = 1000
glyph_names = ['.notdef'] + [f'g{ord(ch):04X}' for ch in CHARACTERS]
fb = FontBuilder(units, isTTF=True)
fb.setupGlyphOrder(glyph_names)
cmap = {ord(ch): f'g{ord(ch):04X}' for ch in CHARACTERS}
fb.setupCharacterMap(cmap)
glyf = {}
# A visible .notdef box.
pen = TTGlyphPen(None)
pen.moveTo((80, 0)); pen.lineTo((620, 0)); pen.lineTo((620, 700)); pen.lineTo((80, 700)); pen.closePath()
pen.moveTo((160, 80)); pen.lineTo((160, 620)); pen.lineTo((540, 620)); pen.lineTo((540, 80)); pen.closePath()
glyf['.notdef'] = pen.glyph()
metrics = {'.notdef': (700, 40)}
for ch in CHARACTERS:
    name = f'g{ord(ch):04X}'
    pen = TTGlyphPen(None)
    if ch != ' ': glyph_outline(glyph_images[ch], pen, units=units)
    glyf[name] = pen.glyph()
    metrics[name] = (350 if ch == ' ' else 700, 40)
fb.setupGlyf(glyf)
fb.setupHorizontalMetrics(metrics)
fb.setupHorizontalHeader(ascent=800, descent=-200)
fb.setupNameTable({' familyName': FONT_FAMILY, 'styleName': 'Regular',
                    'uniqueFontIdentifier': FONT_FAMILY + ' Regular 1.0',
                    'fullName': FONT_FAMILY + ' Regular', 'psName': 'ArmenianAIPrototype-Regular',
                    'version': 'Version 1.0'})
fb.setupOS2(sTypoAscender=800, sTypoDescender=-200, usWinAscent=800, usWinDescent=200)
fb.setupPost()
fb.setupMaxp()
fb.save(FONT_FILENAME)
print('Created', FONT_FILENAME.resolve(), '(', FONT_FILENAME.stat().st_size, 'bytes )')
```

```python
# Render a quick specimen with the newly built font.
preview_font = ImageFont.truetype(str(FONT_FILENAME), 52)
sample = 'Հայերեն տառատեսակ
ԱԲԳԴԵԶԷԸԹԺԻԼԽԾԿՀՁՂՃՄՅՆՇՈՉՊՋՌՍՎՏՐՑՒՓՔՕՖ'
preview = Image.new('RGB', (1100, 190), 'white')
ImageDraw.Draw(preview).text((20, 15), sample, font=preview_font, fill='black', spacing=14)
display(preview)
print('TTF:', FONT_FILENAME.resolve())
```
